## Lab 5: Spatial Filtering – Smoothing and Sharpening

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Task 1: 7x7 Box Filter

```python
image = cv2.cvtColor(cv2.imread('images/astronaut.png'), cv2.COLOR_BGR2RGB)
kernel = np.ones((7, 7), np.float32) / 49
smoothed = cv2.filter2D(image, -1, kernel)

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(smoothed)
ax[1].set_title("7x7 Box Filter")
ax[1].axis("off")
plt.show()
```

## Task 2: 5x5 and 21x21 Gaussian Filters

```python
image = cv2.cvtColor(cv2.imread('images/chelsea.png'), cv2.COLOR_BGR2RGB)
gauss_5 = cv2.GaussianBlur(image, (5, 5), 0)
gauss_21 = cv2.GaussianBlur(image, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[1].imshow(gauss_5)
ax[1].set_title("Gaussian 5x5")
ax[2].imshow(gauss_21)
ax[2].set_title("Gaussian 21x21")
for a in ax:
    a.axis("off")
plt.show()
```

## Task 3: 3x3 Laplacian Sharpening

```python
image = cv2.cvtColor(cv2.imread('images/coffee.png'), cv2.COLOR_BGR2RGB)
image_float = image.astype(np.float32) / 255.0
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=1)
sharpened = np.clip(image_float - laplacian, 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[1].imshow(np.clip(laplacian * 0.5 + 0.5, 0, 1))
ax[1].set_title("Laplacian Filter")
ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
for a in ax:
    a.axis("off")
plt.show()
```
